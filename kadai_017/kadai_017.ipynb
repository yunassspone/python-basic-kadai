```python
class Human:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def check_adult(self):
    if self.age >= 20:
      print(f"{self.name}は大人である")
    else:
      print(f"{self.name}は大人ではない")

#インスタンスを複数作成
personA = Human("Aさん", 15)
personB = Human("Bさん", 23)
personC = Human("Cさん", 30)
personD = Human("Dさん", 18)
personE = Human("Eさん", 20)

#リストに追加
person_info = [personA, personB, personC, personD, personE]

#リストの要素文だけcheck_adultメソッドを呼びだす
for i in person_info:
  i.check_adult()
```
